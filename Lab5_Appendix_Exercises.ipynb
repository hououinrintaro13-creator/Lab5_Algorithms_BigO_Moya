```python
# [CELL 1] SETUP & STUDENT CONSTANTS
SEED_DIGIT = 8
ID_SUM = 21
NAME_LENGTH = 4
ARTIST_SUM = 120  # Sum of letters A=1 to Z=26 for chosen artist
```

```python
### **[CELL 2] GENERATE ORIGINAL AND ROTATED INCIDENT RECORDS**

original_records = [10 + i * 5 for i in range(15)]

# 2. Determine rotation point using SEED_DIGIT
rotation_point = SEED_DIGIT % len(original_records)

# 3. Create rotated incident records list
rotated_records = original_records[rotation_point:] + original_records[:rotation_point]

# 4. Generate target incident using ID_SUM and ARTIST_SUM
target_incident = original_records[(ID_SUM + ARTIST_SUM) % len(original_records)]

print("Original Records:", original_records)
print(f"Rotation Point (SEED_DIGIT): {rotation_point}")
print("Rotated Records: ", rotated_records)
print("Target Incident: ", target_incident)
```

```python
# [CELL 3] ROTATED SEARCH & COMPARISON
def search_rotated(data, target):
    low, high = 0, len(data) - 1
    comparisons = 0
    
    while low <= high:
        mid = (low + high) // 2
        comparisons += 1
        
        if data[mid] == target:
            return mid, comparisons
            
        # Check if left half is sorted
        if data[low] <= data[mid]:
            comparisons += 1
            if data[low] <= target < data[mid]:
                high = mid - 1
            else:
                low = mid + 1
        else: # Right half is sorted
            comparisons += 1
            if data[mid] < target <= data[high]:
                low = mid + 1
            else:
                high = mid - 1
                
    return -1, comparisons

def linear_search_records(data, target):
    comparisons = 0
    for i, val in enumerate(data):
        comparisons += 1
        if val == target:
            return i, comparisons
    return -1, comparisons

# Execute searches
rot_idx, rot_comps = search_rotated(rotated_records, target_incident)
lin_idx, lin_comps = linear_search_records(rotated_records, target_incident)

print(f"Rotated Search -> Index: {rot_idx}, Comparisons: {rot_comps}")
print(f"Linear Search  -> Index: {lin_idx}, Comparisons: {lin_comps}")
print(f"Comparison Difference: Linear used {lin_comps - rot_comps} more comparisons.")
```

```python
# [CELL 4] UNIVERSITY ENROLLMENT DATASETS SETUP
# Dataset A: Generated using SEED_DIGIT and ID_SUM
dataset_a = [SEED_DIGIT + i * 3 for i in range(ID_SUM // 2)]

# Dataset B: Generated using NAME_LENGTH and artist parameters
artist_length = 6  # Length of chosen artist's name string
dataset_b = [NAME_LENGTH + i * 4 for i in range(artist_length * 2)]

# Ensure both datasets are sorted in ascending order
dataset_a.sort()
dataset_b.sort()

print("Department A Dataset:", dataset_a)
print("Department B Dataset:", dataset_b)
```

```python
# [CELL 5] CONSOLIDATE SORTED ENROLLMENT RECORDS
def merge_sorted_collections(list_a, list_b):
    merged = []
    i = j = 0
    comparisons = 0
    
    while i < len(list_a) and j < len(list_b):
        comparisons += 1
        if list_a[i] < list_b[j]:
            merged.append(list_a[i])
            i += 1
        else:
            merged.append(list_b[j])
            j += 1
            
    # Append remaining elements
    merged.extend(list_a[i:])
    merged.extend(list_b[j:])
    
    return merged, comparisons

consolidated_dataset, total_comparisons = merge_sorted_collections(dataset_a, dataset_b)

print("Consolidated Dataset:   ", consolidated_dataset)
print(f"Records Before Merge:    {len(dataset_a)} (Dept A) + {len(dataset_b)} (Dept B) = {len(dataset_a) + len(dataset_b)}")
print(f"Records After Merge:     {len(consolidated_dataset)}")
print(f"Comparison Count:        {total_comparisons}")

# Verify sorting order
is_ordered = all(consolidated_dataset[k] <= consolidated_dataset[k+1] for k in range(len(consolidated_dataset) - 1))
print(f"Ordering Verification:   {'Passed (Properly Sorted)' if is_ordered else 'Failed'}")
print("Time Complexity:         O(n + m)")
print("Space Complexity:        O(n + m)")
```

```python
# [CELL 6] E-COMMERCE ORDER PROCESSING COMPARISON
import time

# Define sorting algorithms with metric tracking
def bubble_sort(data):
    a, comparisons, swaps = data.copy(), 0, 0
    n = len(a)
    for i in range(n):
        swapped = False
        for j in range(0, n - i - 1):
            comparisons += 1
            if a[j] > a[j + 1]:
                a[j], a[j + 1] = a[j + 1], a[j]
                swaps += 1
                swapped = True
        if not swapped:
            break
    return a, comparisons, swaps

def selection_sort(data):
    a, comparisons, swaps = data.copy(), 0, 0
    for i in range(len(a) - 1):
        minimum = i
        for j in range(i + 1, len(a)):
            comparisons += 1
            if a[j] < a[minimum]:
                minimum = j
        if minimum != i:
            a[i], a[minimum] = a[minimum], a[i]
            swaps += 1
    return a, comparisons, swaps

def insertion_sort(data):
    a, comparisons, shifts = data.copy(), 0, 0
    for i in range(1, len(a)):
        key, j = a[i], i - 1
        while j >= 0:
            comparisons += 1
            if a[j] <= key:
                break
            a[j + 1] = a[j]
            shifts += 1
            j -= 1
        a[j + 1] = key
    return a, comparisons, shifts

# Generate input conditions based on student parameters
base_size = 40 + ID_SUM
sorted_input = list(range(base_size))
reverse_input = list(range(base_size, 0, -1))
random_input = [(ID_SUM * 7 + i * 13) % 500 for i in range(base_size)]

conditions = {
    "Sorted": sorted_input,
    "Reverse-Ordered": reverse_input,
    "Random": random_input
}

algorithms = {
    "Bubble Sort": bubble_sort,
    "Selection Sort": selection_sort,
    "Insertion Sort": insertion_sort
}

# Run comparisons across conditions
print(f"{'Condition':<16} | {'Algorithm':<16} | {'Comparisons':<12} | {'Movements':<12}")
print("-" * 64)

for cond_name, data in conditions.items():
    for algo_name, algo_func in algorithms.items():
        _, comps, moves = algo_func(data)
        print(f"{cond_name:<16} | {algo_name:<16} | {comps:<12} | {moves:<12}")
    print("-" * 64)
```
